# Experiment 05
## Feature Extraction and Visualization Using CNN

**Dataset:** MNIST (1,000 training images and 200 testing images)

This notebook builds a lightweight CNN, trains it on a small subset of MNIST, extracts feature maps from two convolutional layers, and visualizes them.

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras import layers, models, Model

print("Libraries imported successfully!")
```

### Cell 2 – Download MNIST Dataset

MNIST is downloaded automatically by TensorFlow if it is not already available.

```python
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

print("Dataset downloaded successfully!")
print("Original Training Data:", x_train.shape)
print("Original Testing Data:", x_test.shape)
```

### Cell 3 – Select a Small Dataset

Use only 1,000 training images and 200 testing images to keep the experiment lightweight.

```python
x_train = x_train[:1000]
y_train = y_train[:1000]

x_test = x_test[:200]
y_test = y_test[:200]

print("Small Dataset Selected!")
print("Training Images:", x_train.shape)
print("Testing Images:", x_test.shape)
```

### Cell 4 – Preprocess Images

Normalize pixel values to the range 0–1 and add the grayscale channel dimension.

```python
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0

x_train = np.expand_dims(x_train, axis=-1)
x_test = np.expand_dims(x_test, axis=-1)

print("Image preprocessing completed!")
print("Input Shape:", x_train.shape)
```

### Cell 5 – Display Sample Images

```python
plt.figure(figsize=(10, 4))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i].squeeze(), cmap='gray')
    plt.title("Digit: " + str(y_train[i]))
    plt.axis('off')

plt.tight_layout()
plt.show()
```

### Cell 6 – Create CNN Model

The model contains two convolutional layers, two max-pooling layers, and dense classification layers.

```python
cnn_model = models.Sequential([
    layers.Input(shape=(28, 28, 1)),

    layers.Conv2D(32, (3, 3), activation='relu', name='conv1'),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation='relu', name='conv2'),
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

cnn_model.summary()

print("CNN Model created successfully!")
```

### Cell 7 – Compile and Train CNN

Train the model for 3 epochs using the selected small MNIST subset.

```python
cnn_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

history = cnn_model.fit(
    x_train,
    y_train,
    epochs=3,
    batch_size=32,
    validation_data=(x_test, y_test)
)

print("CNN Training completed successfully!")
```

### Cell 8 – Extract Features from CNN Layers

Create a feature-extraction model using the outputs of the two convolutional layers. The model inputs are referenced through `cnn_model.inputs`.

```python
feature_model = Model(
    inputs=cnn_model.inputs,
    outputs=[
        cnn_model.get_layer('conv1').output,
        cnn_model.get_layer('conv2').output
    ]
)

sample_image = x_test[0:1]

feature_maps = feature_model.predict(sample_image)

print("Feature extraction completed!")

print("First Layer:", feature_maps[0].shape)
print("Second Layer:", feature_maps[1].shape)
```

### Cell 9 – Visualize Extracted Feature Maps

Display the first eight feature maps from each convolutional layer.

```python
for layer_index, feature_map in enumerate(feature_maps):

    plt.figure(figsize=(12, 5))
    plt.suptitle("CNN Feature Maps - Layer " + str(layer_index + 1))

    for i in range(8):
        plt.subplot(2, 4, i + 1)
        plt.imshow(feature_map[0, :, :, i], cmap='viridis')
        plt.axis('off')

    plt.tight_layout()
    plt.show()

print("Feature map visualization completed!")
```

### Cell 10 – Display Original Image and Extracted Features

```python
plt.figure(figsize=(10, 4))

plt.subplot(1, 3, 1)
plt.imshow(sample_image[0].squeeze(), cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(feature_maps[0][0, :, :, 0], cmap='viridis')
plt.title("Conv Layer 1")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(feature_maps[1][0, :, :, 0], cmap='viridis')
plt.title("Conv Layer 2")
plt.axis('off')

plt.tight_layout()
plt.show()

print("Experiment 05 Completed Successfully!")
```

## Result

Successfully implemented CNN-based feature extraction using the MNIST dataset. Feature maps were extracted from two convolutional layers and visualized using Matplotlib.
